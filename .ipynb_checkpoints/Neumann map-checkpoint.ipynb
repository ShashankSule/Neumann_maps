# Make four cluster data

```python
# each point is a column vector
d = 2 
N = 100
N_cluster = 5 
sig = 0.1*np.eye(d)
means = [np.array([1.0,0.0]), np.array([0.0,1.0]), np.array([-1.0,0.0]), np.array([0.0,-1.0]), np.array([0.0,0.0])]
# means = [means[i].reshape((2,1)) for i in range(N_cluster)]
```

```python
# data 

data = []
labels = [] 
for i in range(N_cluster): 
    cluster = np.random.multivariate_normal(means[i].flatten(), sig, N).T
    label = i*np.ones(N)
    data.append(cluster)
    labels.append(label)
data = np.hstack(data)
labels = np.hstack(labels)
```

```python
# plot 
plt.scatter(data[0,:], data[1,:], c=labels, cmap='jet')
```

# Construct NMap

```python
# get dmap
eps = 0.25
alpha = 0.5
nmap = diffusion_map.DiffusionMap(alpha=alpha, \
                                  epsilon=eps, \
                                  num_evecs=2)
```

```python
# construct kernel 
K = nmap._construct_kernel(data)

# alpha renorm 
D_alph_inv_vec = (K@np.ones((K.shape[1],1)).flatten())**(-alpha)
D_alph_inv = scipy.sparse.diags(D_alph_inv_vec)
K = D_alph_inv@K@D_alph_inv

# construct graph laplacian
D = scipy.sparse.diags(K@np.ones((K.shape[1],1)).flatten())
L = D - K
```

```python
# construct boundary info 

subgraph_inds = ~(labels == 4.0)
# K_sub = K[subgraph_inds,:][:,subgraph_inds]
# K_boundary = K[~subgraph_inds,:][:,~subgraph_inds]
```

```python
# construct Neumann matrix 

B = K[~subgraph_inds, :][:, subgraph_inds]
delta_TS_vec = B@np.ones((B.shape[1],1)).flatten()
delta_TS_vec_inv = (1/delta_TS_vec)
delta_TS_inv = scipy.sparse.diags(delta_TS_vec_inv)
L_N = L[subgraph_inds, :][:, subgraph_inds] - B.T@delta_TS_inv@B
T_S = D.tocsr()[subgraph_inds,:][:,subgraph_inds]
K_N = T_S - L_N
```

```python
# renormalize the kernel matrix

one_over_T_S_sqrt_vec = 1/(K_N@np.ones((K_N.shape[1],1)).flatten())**(1/2)
one_over_T_S_sqrt = scipy.sparse.diags(one_over_T_S_sqrt_vec)
renormalized_K_N = one_over_T_S_sqrt@K_N@one_over_T_S_sqrt
T_S_sqrt_vec = (K_N@np.ones((K_N.shape[1],1)).flatten())**(1/2)
T_S_sqrt = scipy.sparse.diags(T_S_sqrt_vec)
P_N = one_over_T_S_sqrt@renormalized_K_N@T_S_sqrt
generator = (P_N - scipy.sparse.eye(P_N.shape[0]))/eps
```

```python
# get eigenvectors! 
generator = (P_N - scipy.sparse.eye(P_N.shape[0]))/eps
evals, evecs = scipy.sparse.linalg.eigs(P_N, 3, which = 'LM')
evals = evals[1:]
evecs = evecs[:,1:] # get rid of the ones vector

transition_matrix_used = True 

if transition_matrix_used: 
    Psi = evecs
else: 
    Psi = one_over_T_S_sqrt.toarray()@evecs

Psi = Psi@np.diag(evals)
standardize = False
if standardize:
    norms = np.diag(1/np.sqrt(np.diag(Psi.T@Psi)))
else: 
    norms = np.eye(2)
dmaps = Psi@norms
```

```python
dmaps, evecs, evals = nmap._construct_diffusion_coords(generator)
```

```python
# plot evex 

plt.scatter(dmaps[:,0], dmaps[:,1], c=labels[subgraph_inds], cmap='jet')
# plt.xlim((-2e-4, 2e-4))
# plt.ylim((-2e-4, 2e-4))
```

# Write class

```python
class NeumannMap(diffusion_map.DiffusionMap): 
    def __init__(self, alpha=0.0, epsilon=1.0, num_evecs=3, marked = False, pbc_dims=None,
                 n_neigh=None, density=None):
        super().__init__(alpha=alpha, epsilon=epsilon,
                         num_evecs=num_evecs, pbc_dims=pbc_dims,
                         n_neigh=n_neigh)
        # marked = boundary, the NMap will always be computed on the unmarked pixels 
        if ~marked: 
            self.marked = marked
        else:
            self.marked = ~marked
    
    def construct_generator(self,data,subgraph): 
        
        # compute map on unmarked points 
        if ~self.marked: 
            subgraph_inds = subgraph
        else: 
            subgraph_inds = ~subgraph
        
        # construct kernel 
        K = self._construct_kernel(data)
        
        # alpha renorm 
        D_alph_inv_vec = (K@np.ones((K.shape[1],1)).flatten())**(-self.alpha)
        D_alph_inv = scipy.sparse.diags(D_alph_inv_vec)
        K = D_alph_inv@K@D_alph_inv


        # construct graph laplacian
        D = scipy.sparse.diags(K@np.ones((K.shape[1],1)).flatten())
        L = D - K
        
        # construct Neumann matrix 
        B = K[~subgraph_inds, :][:, subgraph_inds] # boundary matrix 
        delta_TS_vec = B@np.ones((B.shape[1],1)).flatten()
        delta_TS_vec_inv = (1/delta_TS_vec) # deltaT_S matrix 
        delta_TS_inv = scipy.sparse.diags(delta_TS_vec_inv)
        L_N = L[subgraph_inds, :][:, subgraph_inds] - B.T@delta_TS_inv@B # neumann laplacian 
        T_S = D.tocsr()[subgraph_inds,:][:,subgraph_inds] # degree matrix of subgraph 
        K_N = T_S - L_N # Neumann kernel matrix 
        
        # renormalize the kernel matrix
        one_over_T_S_sqrt_vec = 1/(K_N@np.ones((K_N.shape[1],1)).flatten())**(1/2)
        one_over_T_S_sqrt = scipy.sparse.diags(one_over_T_S_sqrt_vec)
        renormalized_K_N = one_over_T_S_sqrt@K_N@one_over_T_S_sqrt
        T_S_sqrt_vec = (K_N@np.ones((K_N.shape[1],1)).flatten())**(1/2)
        T_S_sqrt = scipy.sparse.diags(T_S_sqrt_vec)
        P_N = one_over_T_S_sqrt@renormalized_K_N@T_S_sqrt # transition matrix of reflecting random walk 
        
        generator = (P_N - scipy.sparse.eye(P_N.shape[0]))/self.epsilon # generator of reflecting walk 
        
        self.L = generator
        
        return self 
```

```python
nmap2 = NeumannMap(alpha=0.5, epsilon=eps, num_evecs=2)
```

```python
nmap2.construct_generator(data,subgraph_inds)
```

```python
L2 = nmap2.get_generator()
```

```python
dmap2, evecs2, evals2 = nmap2._construct_diffusion_coords(L2)
```
